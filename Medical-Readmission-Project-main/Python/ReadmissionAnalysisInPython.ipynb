```python
#Load Data From CSV
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df = pd.read_csv('AverageReadmits.csv')
```

```python
"""Show the first five rows to ensure things loaded properly"""
print(df.head())
"""The shape of the data is 12 rows one for each month with 2 collumns with one being the month and the second being the readmit rate."""
print(df.shape)
```

```python
x,y = df['Service_Month_Name'],df['Avg_Readmission_Rate']
```

```python
"""I wanted to plot the data for average readmission rates per month because I wanted to see if it was improving or getting worse."""

x = df['Service_Month_Name']
y = df['Avg_Readmission_Rate']

x_num = np.arange(len(y))

r = np.corrcoef(x_num, y)[0, 1]
r2 = r**2

plt.figure(figsize=(8, 5))
plt.plot(
    x,
    y,
    marker='o',
    linestyle='-',
    linewidth=2,
    label='Avg Readmission Rate'
)

plt.xlabel("Service Month")
plt.ylabel("Average Readmission Rate")
plt.title("Average Readmission Rate Over Time")

plt.xticks(rotation=45)
plt.text(
    0.02,
    0.95,
    f"r = {r:.3f}\nR² = {r2:.3f}",
    transform=plt.gca().transAxes,
    verticalalignment="top",
    bbox=dict(boxstyle="round", alpha=0.3)
)

plt.legend()
plt.tight_layout()
plt.show()
```
